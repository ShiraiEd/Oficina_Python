## 1. Comparações viram True ou False
`True` e `False` são o tipo `bool`.

![Tabela dos comparadores ==, !=, >, <, >= e <= com exemplos e resultados](../imagens/comparadores.png)

```python
print(7 >= 5)
print(3 > 4)
print(10 == 10)
print(10 != 10)
```

## 2. and, or, not
- `and`: as **duas** condições precisam ser verdadeiras
- `or`: basta **uma** ser verdadeira
- `not`: inverte (`True` vira `False`)

![Tabela verdade de A and B, A or B e not A para todas as combinações de True e False](../imagens/tabela-verdade.png)

```python
nota = 8
frequencia = 70

print(nota >= 7 and frequencia >= 75)
print(nota >= 7 or frequencia >= 75)
print(not (nota >= 7))
```

**EXPERIMENTE:** mude a frequência para `80`. O que muda nas três linhas?

## 3. Usando no if

```python
nota = 8
frequencia = 70

if nota >= 7 and frequencia >= 75:
    print("Aprovado")
else:
    print("Reprovado")
```

## Preveja antes de rodar
Com `x = 15`, o que aparece em cada linha? Combine no grupo e **depois** rode.

```python
x = 15
print(x > 10 and x < 20)
print(x > 10 and x < 12)
print(x < 10 or x == 15)
print(not (x == 15))
```
