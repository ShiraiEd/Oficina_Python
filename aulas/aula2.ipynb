## 1. if / else
O programa escolhe um caminho dependendo de uma condição.

Repare nos **dois-pontos** no fim da linha e nos **4 espaços** antes do `print`. É assim que o Python sabe o que está dentro do `if`.

```python
numero_secreto = 42
palpite = 50

if palpite == numero_secreto:
    print("Acertou!")
else:
    print("Errou!")
```

**Cuidado:** `=` guarda um valor; `==` compara dois valores.

**EXPERIMENTE:** mude o palpite para `42` e rode de novo.

## 2. elif
`elif` permite mais de duas opções. O programa testa de cima para baixo e segue **só um** caminho:

![Fluxograma: se o palpite é igual ao número secreto, mostra Acertou; senão, se é maior, mostra Muito alto; senão, mostra Muito baixo](../imagens/fluxo-if.png)

```python
numero_secreto = 42
palpite = 50

if palpite == numero_secreto:
    print("Acertou!")
elif palpite > numero_secreto:
    print("Muito alto!")
else:
    print("Muito baixo!")
```

**EXPERIMENTE:**
- Teste com palpite `10`, `42` e `99`.
- Tire os 4 espaços antes de um `print` e rode. Leia o erro (`IndentationError`) e depois conserte.

## 3. O palpite vindo do jogador

```python
numero_secreto = 42
palpite = int(input("Seu palpite: "))

if palpite == numero_secreto:
    print("Acertou!")
elif palpite > numero_secreto:
    print("Muito alto!")
else:
    print("Muito baixo!")
```
