## 1. while
O `while` repete o bloco **enquanto** a condição for verdadeira.

```python
contador = 1

while contador <= 3:
    print("Contando:", contador)
    contador = contador + 1
```

**Cuidado:** se esquecer de mudar o `contador`, o laço nunca termina.
Para parar uma célula travada, clique no botão **■ (Interromper)** ao lado da célula.

**EXPERIMENTE:** mude o `3` para `5`.

## 2. for + range()
O `for` também repete código, mas geralmente é usado quando você já sabe quantas vezes quer repetir.

`range()` gera uma sequência de números. Por exemplo:

- `range(5)` -> 0, 1, 2, 3, 4
- `range(1, 6)` -> 1, 2, 3, 4, 5
- `range(5, 0, -1)` -> 5, 4, 3, 2, 1

Isso é muito útil para contagens e contagem regressiva.

```python
for numero in range(1, 6):
    print(numero)
```

## 3. O jogo repetindo
Agora o jogo continua até o jogador acertar. Repare na seta azul: depois do bloco, o programa **volta** e testa a condição de novo.

![Fluxograma do while: enquanto não acertou, pede um palpite e confere; quando acertar, sai do laço e termina o jogo](../imagens/fluxo-while.png)

Teste com `50`, depois `10`, depois `42`.

```python
numero_secreto = 42
acertou = False

while not acertou:
    palpite = int(input("Seu palpite: "))

    if palpite == numero_secreto:
        print("Acertou!")
        acertou = True
    elif palpite > numero_secreto:
        print("Muito alto!")
    else:
        print("Muito baixo!")
```

**EXPERIMENTE:** mude o número secreto. O jogo continua funcionando?
