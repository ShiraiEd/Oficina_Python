## 1. print e variáveis
Uma variável é uma caixinha com nome que guarda um valor.

```python
nome = "Ana"
idade = 20

print("Olá,", nome)
print("Ano que vem você terá", idade + 1, "anos")
```

**EXPERIMENTE:** troque o nome e a idade e rode de novo.

## 2. Tipos
Todo valor tem um **tipo**, e o tipo define o que dá para fazer com ele.

![Tabela com os tipos str, int, float e bool, o que cada um guarda, exemplos e a função para converter](../imagens/tipos.png)

```python
nome = "Ana"       # str   (texto)
idade = 20         # int   (número inteiro)
altura = 1.65      # float (número com casas decimais)

print(type(nome))
print(type(idade))
print(type(altura))
```

O mesmo `+` faz coisas diferentes dependendo do tipo:

```python
print("5" + "3")   # texto: junta
print(5 + 3)       # número: soma
```

**EXPERIMENTE:** crie uma célula nova com `print("5" + 3)` e rode. Leia a última linha do erro (`TypeError`): o Python não sabe somar texto com número.

## 3. input e conversão
`input()` pergunta algo ao usuário. A caixa de digitação aparece **no topo da tela**.

**Cuidado:** `input()` *sempre* devolve texto (`str`), mesmo que você digite um número.

```python
x = input("Digite um número: ")
print(x + x)
```

Se digitou `4`, apareceu `44`: dois textos juntos. Para virar número, convertemos com `int()` (inteiro), `float()` (decimal) ou `str()` (texto).

```python
x = int(x)
print(x + x)
```

**Erro comum:** `ValueError: invalid literal for int()` significa que o `int()` recebeu algo que não é um número inteiro (letras, vazio, `3.5`).
