## 1. import
O Python vem com "caixas de ferramentas" prontas, chamadas **módulos**. O `import` liga uma delas. Sem ele, o Python não sabe o que é `random`.

`random.randint(1, 6)` quer dizer: "da caixa `random`, use a ferramenta `randint`", que sorteia um inteiro de 1 a 6 (**os dois extremos entram**).

```python
import random

print(random.randint(1, 6))   # um dado
```

**EXPERIMENTE:** rode a célula várias vezes. O número muda?

Se aparecer `NameError: name 'random' is not defined`, faltou o `import`.

## 2. O jogo completo
Agora o número secreto é sorteado e o jogo conta as tentativas. Palpites fora do intervalo não contam.

```python
import random

limite = 100
numero_secreto = random.randint(1, limite)
tentativas = 0
acertou = False

nome = input("Qual é o seu nome? ")
print("Bem-vindo ao jogo,", nome)
print("Estou pensando em um número de 1 a", limite)

while not acertou:
    palpite = int(input("Seu palpite: "))

    if palpite < 1 or palpite > limite:
        print("Palpite fora do intervalo! Esse não conta como tentativa.")
    else:
        tentativas = tentativas + 1

        if palpite == numero_secreto:
            print("Acertou em", tentativas, "tentativas!")
            acertou = True
        elif palpite > numero_secreto:
            print("Muito alto!")
        else:
            print("Muito baixo!")

print("Fim de jogo. Parabéns,", nome)
```

**EXPERIMENTE:**
- Mude `limite` para `10`. O jogo fica mais fácil?
- Mude `limite` para `1000`. Quantas tentativas você precisa?
