## Exercício 2 (desafio)
Limite o jogo a **7 tentativas**. Se acabarem, mostre `Acabaram as tentativas! O número era 42`.

Aqui o número secreto é **fixo** (42) para o `checar` conseguir testar.

Dicas:
- Mude a condição do `while` para também parar quando `tentativas` chegar a 7.
- Depois do `while`, use um `if` para avisar quem não acertou.

````python
limite = 100
numero_secreto = 42
tentativas = 0
acertou = False

while not acertou:
    palpite = int(input("Seu palpite: "))

    if palpite < 1 or palpite > limite:
        print("Palpite fora do intervalo! Esse não conta como tentativa.")
    else:
        tentativas = tentativas + 1

        if palpite == numero_secreto:
            print("Acertou em", tentativas, "tentativas!")
            acertou = True
        elif palpite > numero_secreto:
            print("Muito alto!")
        else:
            print("Muito baixo!")

# escreva aqui o aviso para quem não acertou
````
